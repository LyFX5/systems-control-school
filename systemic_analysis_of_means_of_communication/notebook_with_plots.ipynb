```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

x = np.array([2, 3])
y = np.array([4, 2])

ax.scatter([x[0]], [y[0]], color='red', marker='o', s=200)
ax.scatter([x[1]], [y[1]], color='blue', marker='o', s=200)

plt.annotate(r"$\bar{x}_1$" + f" = ({x[0]}, {y[0]})", # this is the text
                 (x[0], y[0]), # these are the coordinates to position the label
                 textcoords="offset points", # how to position the text
                 xytext=(0,10), # distance from text to points (x,y)
                 ha='center', # horizontal alignment can be left, right or center
             fontsize=15) 

plt.annotate(r"$\bar{x}_2$" + f" = ({x[1]}, {y[1]})", # this is the text
                 (x[1], y[1]), # these are the coordinates to position the label
                 textcoords="offset points", # how to position the text
                 xytext=(0,10), # distance from text to points (x,y)
                 ha='center', # horizontal alignment can be left, right or center
             fontsize=15) 

ax.set_xlabel('Расстояние между \n первым и вторым витками \n (первая переменная состояния)', fontsize=13)
ax.set_ylabel('Расстояние между \n вторым и третьим витками \n (вторая переменная состояния)', fontsize=13)

plt.xlim([0, 5])
plt.ylim([0, 5])
plt.title("Фазовое Пространство Пружины \n с Двумя Переменными Состояния \n", fontsize=15)
plt.grid()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

np.random.seed(491)

x1 = np.random.randint(8, size=5) + 1
y1 = np.random.randint(8, size=5) + 1

x2 = np.random.randint(8, size=3) + 1
y2 = np.random.randint(8, size=3) + 1

# ax.scatter(x1, y1, color='red', marker='o', s=100)
# ax.scatter(x2, y2, color='blue', marker='o', s=100)

ax.scatter(x1, y1, color='black', marker='o', s=100)
ax.scatter(x2, y2, color='black', marker='o', s=100)

for i in range(5):
    plt.annotate(f"состояние {i+1}: " + "{0:b}".format(i), # this is the text
             (x1[i], y1[i]), # these are the coordinates to position the label
             textcoords="offset points", # how to position the text
             xytext=(0,10), # distance from text to points (x,y)
             ha='center', # horizontal alignment can be left, right or center
         fontsize=7) 
        
for j in range(3):
    plt.annotate(f"состояние {i+j+2}: " + "{0:b}".format(i+j+1), # i+j+1 this is the text
             (x2[j], y2[j]), # these are the coordinates to position the label
             textcoords="offset points", # how to position the text
             xytext=(0,10), # distance from text to points (x,y)
             ha='center', # horizontal alignment can be left, right or center
         fontsize=7) 


ax.set_xlabel(r'Переменная состояния $x_1$', fontsize=13)
ax.set_ylabel(r'Переменная состояния $x_2$', fontsize=13)

plt.xlim([0, 10])
plt.ylim([0, 10])
plt.title("Двухмерное Фазовое Пространство \n некоторой системы \n", fontsize=15)
plt.grid()
plt.show()
```

```python
a = 0
b = 100
```

```python
def my_density_uniform(x):
    return np.ones_like(x) * (1 / (b-a+1))
```

```python
def my_density_normal(x, mu, sigma_sq):
    return (1 / (2 * np.pi * sigma_sq)**0.5) * np.exp(-(x - mu)**2 / (2 * sigma_sq))
```
